# ChatGPT AI Signals: Algeria check

This notebook extracts the CSV files in `data/ai/data-download-csv` and checks whether the country ranking data include Algeria.

## Setup

```python
from pathlib import Path

import pandas as pd
```

```python
def find_project_root(marker="pyproject.toml"):
    """Walk up from the current directory until the project root is found."""
    current = Path.cwd()
    for parent in [current, *current.parents]:
        if (parent / marker).exists():
            return parent
    raise FileNotFoundError(f"Could not find {marker} in any parent directory")


PROJECT_ROOT = find_project_root()
AI_DATA_DIR = PROJECT_ROOT / "data" / "ai" / "data-download-csv"

COUNTRY_CODE = "DZ"
COUNTRY_NAME = "Algeria"
```

## Extract CSV Files

```python
csv_files = sorted(AI_DATA_DIR.glob("*.csv"))

csv_inventory = pd.DataFrame(
    {
        "file_name": [path.name for path in csv_files],
        "path": [path.relative_to(PROJECT_ROOT).as_posix() for path in csv_files],
    }
)

csv_inventory
```

## Check For Algeria

```python
country_rank = pd.read_csv(
    AI_DATA_DIR / "share_of_messages_by_country_quarter_rank.csv",
    parse_dates=["quarter"],
)

algeria_rank = (
    country_rank.loc[lambda df: df["country"].eq(COUNTRY_CODE)]
    .sort_values("quarter")
    .reset_index(drop=True)
)

print(f"Contains {COUNTRY_NAME}: {not algeria_rank.empty}")
algeria_rank
```

## Latest Quarter

```python
latest_quarter = country_rank["quarter"].max()
latest_country_rank = country_rank.loc[lambda df: df["quarter"].eq(latest_quarter)]
latest_algeria_rank = latest_country_rank.loc[
    lambda df: df["country"].eq(COUNTRY_CODE)
].reset_index(drop=True)

print(f"Latest quarter: {latest_quarter.date()}")
latest_algeria_rank
```
