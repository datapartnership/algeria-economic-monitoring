# Future of Business Survey - Algeria Analysis

## About the Future of Business Survey

The **Future of Business Survey** is a collaborative initiative between Facebook, the Organisation for Economic Co-operation and Development (OECD), and the World Bank, designed to provide timely insights into the perceptions, challenges, and outlook of online Small and Medium Enterprises (SMEs).

### Key Features:
- **Launch**: First launched in February 2016 with 22 countries, now covering over 97 countries globally
- **Frequency**: Initially monthly (2016-2018), moved to bi-annual cadence in 2019
- **Target Population**: SMEs with active Facebook business pages
- **Survey Topics**:
  - Perceptions of current and future economic activity
  - Business challenges and outlook
  - Business characteristics and strategy
  - Custom modules on regulation, access to finance, digital payments, and digital skills

### Sampling Procedure

The survey implements a **modified simple random sample** for invitations within each country:

1. **Sampling Frame**: All Facebook accounts that are admins of active Facebook Pages
2. **Random Assignment**: All Facebook accounts, including Page admins, are randomly assigned to general survey eligibility
3. **Selection Process**: A random sample of firms representing the target population in each country is selected for each survey wave
4. **Ineligibility Period**: Once a respondent has been invited to a survey (regardless of response), they are ineligible for another survey for up to six months. This reduces response burden and maintains response quality
5. **Biannual Cadence**: The survey is fielded biannually to ensure the population from wave to wave remains as consistent as possible
6. **Weighting**: Results are weighted back to the base population using survey weights, based on known characteristics of the Facebook Page admin population

All responses are anonymized, and only country-level data is shared publicly.

### Algeria Data Coverage

This analysis examines Algeria's participation in the Future of Business Survey across 5 survey waves. 

**Data Availability:**
- **2018**: 1 survey wave (Winter)
- **2019**: 2 survey waves (Spring, Winter)
- **2020**: No Algeria data available
- **2021**: 1 survey wave (September)
- **2022**: 1 survey wave (March)

---

**Data Source**: [World Bank Microdata Library - Future of Business Survey](https://microdata.worldbank.org/index.php/catalog/4212)

```python

# Filter for Algeria only (ISO-2: DZ, ISO-3: DZA)
def filter_algeria(df_data, df_name):
    """Filter dataframe to only include Algeria data"""
    original_shape = df_data.shape
    
    # Check for ISO2 or ISO3 columns first
    iso2_cols = [col for col in df_data.columns if 'iso2' in col.lower() or 'iso_2' in col.lower()]
    iso3_cols = [col for col in df_data.columns if 'iso3' in col.lower() or 'iso_3' in col.lower()]
    country_cols = [col for col in df_data.columns if 'country' in col.lower()]
    
    if iso2_cols:
        iso_col = iso2_cols[0]
        df_filtered = df_data[df_data[iso_col] == 'DZ'].reset_index(drop=True)
        print(f"{df_name}: {original_shape} -> {df_filtered.shape} (filtered for Algeria using {iso_col})")
        return df_filtered
    elif iso3_cols:
        iso_col = iso3_cols[0]
        df_filtered = df_data[df_data[iso_col] == 'DZA'].reset_index(drop=True)
        print(f"{df_name}: {original_shape} -> {df_filtered.shape} (filtered for Algeria using {iso_col})")
        return df_filtered
    elif country_cols:
        country_col = country_cols[0]
        df_filtered = df_data[df_data[country_col].isin(['DZ', 'DZA', 'Algeria'])].reset_index(drop=True)
        print(f"{df_name}: {original_shape} -> {df_filtered.shape} (filtered for Algeria using {country_col})")
        return df_filtered
    else:
        print(f"{df_name}: {original_shape} (no ISO or country column found - keeping all data)")
        return df_data

# Apply filtering to all dataframes
fob19_spring = filter_algeria(df, 'fob19_spring')
fob19_winter = filter_algeria(fob19_winter, 'fob19_winter')
fob20_may = filter_algeria(fob20_may, 'fob20_may')
fob20_june = filter_algeria(fob20_june, 'fob20_june')
fob20_july = filter_algeria(fob20_july, 'fob20_july')
fob20_aug = filter_algeria(fob20_aug, 'fob20_aug')
fob20_sep = filter_algeria(fob20_sep, 'fob20_sep')
fob20_oct = filter_algeria(fob20_oct, 'fob20_oct')
fob21_feb = filter_algeria(fob21_feb, 'fob21_feb')
fob21_jul = filter_algeria(fob21_jul, 'fob21_jul')
fob21_sep = filter_algeria(fob21_sep, 'fob21_sep')
fob22_jan = filter_algeria(fob22_jan, 'fob22_jan')
fob22_mar = filter_algeria(fob22_mar, 'fob22_mar')
fob22_july = filter_algeria(fob22_july, 'fob22_july')

print("\n" + "=" * 80)
print("ALGERIA DATA SUMMARY:")
print(f"fob18_winter: {fob18.shape} (already filtered)")
print(f"fob19_spring: {fob19_spring.shape}")
print(f"fob19_winter: {fob19_winter.shape}")
print(f"fob20_may: {fob20_may.shape}")
print(f"fob20_june: {fob20_june.shape}")
print(f"fob20_july: {fob20_july.shape}")
print(f"fob20_aug: {fob20_aug.shape}")
print(f"fob20_sep: {fob20_sep.shape}")
print(f"fob20_oct: {fob20_oct.shape}")
print(f"fob21_feb: {fob21_feb.shape}")
print(f"fob21_jul: {fob21_jul.shape}")
print(f"fob21_sep: {fob21_sep.shape}")
print(f"fob22_jan: {fob22_jan.shape}")
print(f"fob22_mar: {fob22_mar.shape}")
print(f"fob22_july: {fob22_july.shape}")
```

```python
# Analyze VARIABLES and QUESTIONS separately (Algeria data only)
from collections import Counter

# Extract variables (var names) and questions (text) separately
all_variables = {}
all_questions = {}

for name, cb in codebooks_dict_filtered.items():
    # Extract VARIABLES (the column/variable names)
    var_col = None
    if 'var' in cb.columns:
        var_col = 'var'
    elif 'variable' in cb.columns:
        var_col = 'variable'
    elif 'Variable' in cb.columns:
        var_col = 'Variable'
    elif 'Column' in cb.columns:
        var_col = 'Column'
    
    if var_col:
        all_variables[name] = set(cb[var_col].dropna().unique())
    else:
        # Use first column if no explicit variable column
        all_variables[name] = set(cb.iloc[:, 0].dropna().unique())
    
    # Extract QUESTIONS (the question text)
    if 'question' in cb.columns:
        all_questions[name] = set(cb['question'].dropna().unique())
    elif 'Question' in cb.columns:
        all_questions[name] = set(cb['Question'].dropna().unique())
    elif 'label' in cb.columns:
        all_questions[name] = set(cb['label'].dropna().unique())
    else:
        all_questions[name] = set()
    
    print(f"{name}: {len(all_variables[name])} variables, {len(all_questions[name])} questions")

print("\n" + "=" * 80)
print("VARIABLES Analysis (Algeria data only):")
print("=" * 80)
common_vars = set.intersection(*all_variables.values())
print(f"Variables common to ALL codebooks with Algeria data: {len(common_vars)}")
if len(common_vars) <= 30:
    for v in sorted(common_vars):
        print(f"  - {v}")

print("\n" + "=" * 80)
print("QUESTIONS Analysis (Algeria data only):")
print("=" * 80)
if all(len(q) > 0 for q in all_questions.values()):
    common_qs = set.intersection(*all_questions.values())
    print(f"Questions common to ALL codebooks with Algeria data: {len(common_qs)}")
    if len(common_qs) <= 10:
        for q in sorted(common_qs):
            print(f"  - {q[:100]}...")
else:
    print("Not all codebooks have question text columns")
```

```python
# Variable frequency analysis (Algeria data only)
var_counter = Counter()
for variables in all_variables.values():
    var_counter.update(variables)

print("\nVARIABLE frequency across codebooks (Algeria data only):")
print("=" * 80)
by_var_freq = {}
for v, count in var_counter.items():
    if count not in by_var_freq:
        by_var_freq[count] = []
    by_var_freq[count].append(v)

for count in sorted(by_var_freq.keys(), reverse=True):
    variables = by_var_freq[count]
    print(f"\nAppears in {count}/{len(codebooks_dict_filtered)} codebooks ({len(variables)} variables):")
    if len(variables) <= 20:
        for v in sorted(variables):
            print(f"  - {v}")
    else:
        print(f"  (Showing first 20):")
        for v in sorted(variables)[:20]:
            print(f"  - {v}")

# Question frequency analysis
print("\n\n" + "=" * 80)
print("QUESTION frequency across codebooks (Algeria data only):")
print("=" * 80)
question_counter = Counter()
for questions in all_questions.values():
    if len(questions) > 0:
        question_counter.update(questions)

if question_counter:
    by_q_freq = {}
    for q, count in question_counter.items():
        if count not in by_q_freq:
            by_q_freq[count] = []
        by_q_freq[count].append(q)
    
    for count in sorted(by_q_freq.keys(), reverse=True):
        questions = by_q_freq[count]
        print(f"\nAppears in {count}/{len(codebooks_dict_filtered)} codebooks ({len(questions)} questions):")
        if len(questions) <= 10:
            for q in sorted(questions):
                print(f"  - {q[:100]}...")
        else:
            print(f"  (Showing first 10):")
            for q in sorted(questions)[:10]:
                print(f"  - {q[:100]}...")
else:
    print("No question text data available in codebooks")
```

```python
# Summary: Total unique variables and questions (Algeria data only)
print("\n" + "=" * 80)
print("SUMMARY (Algeria data only):")
print("=" * 80)
print(f"Total unique VARIABLES across codebooks with Algeria data: {len(var_counter)}")
print(f"Total unique QUESTIONS across codebooks with Algeria data: {len(question_counter)}")
print(f"Total codebooks analyzed (with Algeria data): {len(codebooks_dict_filtered)}")

print(f"\nVARIABLE breakdown by frequency:")
for count in sorted(by_var_freq.keys(), reverse=True):
    print(f"  {count} codebooks: {len(by_var_freq[count])} variables")
    

if question_counter:
    print(f"\nQUESTION breakdown by frequency:")
    for count in sorted(by_q_freq.keys(), reverse=True):
        print(f"  {count} codebooks: {len(by_q_freq[count])} questions")
```
